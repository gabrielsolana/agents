# And now - TOOLS!

Let's start with a function...

````python
def record_email_tool(email):
    print(f"Tool called to record an email: {email}")
    with open("emails.txt", "a", encoding="utf-8") as f:
        f.write(email + "\n")
    return "Email received"
````

````python
record_email_tool("gsol@iestiemposmodernos.com")
````

## Step 1 - write some json to describe the tool


````python
record_email_tool_json = {
    "name": "record_email_tool",
    "description": "Use this tool to record that a user provided their email address",
    "parameters": {
        "type": "object",
        "properties": {
            "email": {"type": "string", "description": "The email address of this user"}
        },
        "required": ["email"],
        "additionalProperties": False
    }
}
````

````python
tools = [{"type": "function", "function": record_email_tool_json}]
````

````python
tools
````

## Step 2 - a new chat() function

This is where we implement the tool call.

The reality is, it's a bit clunky. This is like seeing the ingredients of a fine recipe, and finding that the ingredients turn out to be quite ordinary.

Tool calling is an "if" statement. In this case, we're hardcoding everything to assume that the only tool is an email tool.

SIDENOTE: If you're thinking - but wait! I should be remembering this so I can do it myself! Then the key point is: this is what Agent Frameworks take care of for you. In practice, you'll likely never type this again yourself. We are shielded from these if statements by the Agent Framework. That's why they're often described as "abstraction layers".

````python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)
         
    if response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            tool_call = message.tool_calls[0]
            email = json.loads(tool_call.function.arguments).get("email")
            record_email_tool(email)
            messages.append(message)
            messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)
            
    return response.choices[0].message.content
````

````python
chat("", [])
````

````python
gr.ChatInterface(chat).launch(inbrowser=True)
````

## Step 3

Our first ever Agent Loop, done without an Agent Framework!

Changes:
1. Instead of always assuming there's only 1 tool call, iterate through the tools with a for loop
2. Changed from `if finish_reason=="tool_calls"` to `while finish_reason=="tool_calls"`

````python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)
         
    while response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            messages.append(message)
            for tool_call in message.tool_calls:
                email = json.loads(tool_call.function.arguments).get("email")
                record_email_tool(email)
                messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)
            
    return response.choices[0].message.content
````

````python
gr.ChatInterface(chat).launch(inbrowser=True)
````

## Ejercicio 1 - Evaluador "solo trabajo"

Paso 1: función `evaluate(reply)` aislada. Una llamada LLM que devuelve JSON `{is_work_related, reason}`.

````python
evaluator_prompt = """You are a strict reviewer of a digital twin's replies.
A reply is work related (true) ONLY if it either:
- talks about the represented person's career, background, skills, experience or projects, or how to get in touch for work; or
- politely declines an off-topic request and redirects to professional topics, WITHOUT giving any off-topic content.
It is NOT work related (false) if it contains off-topic content (recipes, sports, trivia...), or offers to help with off-topic things,
even if it uses professional wording (e.g. "turn the recipe into a professional technical sheet").
Always answer by using your tool to record the evaluation."""

record_evaluation_json = {
    "name": "record_evaluation",
    "description": "Use this tool to record whether a reply is strictly work related",
    "parameters": {
        "type": "object",
        "properties": {
            "is_work_related": {"type": "boolean", "description": "True only if the reply is strictly work related"},
            "reason": {"type": "string", "description": "One short sentence explaining the verdict"}
        },
        "required": ["is_work_related", "reason"],
        "additionalProperties": False
    }
}

evaluator_tools = [{"type": "function", "function": record_evaluation_json}]

def evaluate(reply):
    messages = [
        {"role": "system", "content": evaluator_prompt},
        {"role": "user", "content": reply},
    ]
    response = openai.chat.completions.create(
        model="gpt-5.4-mini", messages=messages, tools=evaluator_tools,
        tool_choice={"type": "function", "function": {"name": "record_evaluation"}},
    )
    return json.loads(response.choices[0].message.tool_calls[0].function.arguments)
````

````python
# Paso 2: prueba del evaluador con casos conocidos
cases = [
    ("Tengo experiencia liderando proyectos de IA y automatización durante 5 años.", True),
    ("Domino Python, SQL y herramientas de agentes como LangGraph.", True),
    ("Para hacer una buena tortilla de patatas necesitas huevos, patatas y cebolla.", False),
    ("Creo que el Real Madrid ganará la liga este año por su plantilla.", False),
    ("Me encanta cocinar, pero lo que de verdad me define profesionalmente es mi trabajo en IA.", True),
]
for reply, expected in cases:
    result = evaluate(reply)
    print("PASS" if result["is_work_related"] == expected else "FAIL", result)
````

## Pasos 3-5 - Nueva `chat()` con evaluación

Tras formar la respuesta (con el bucle de tools), `evaluate()` comprueba que sea solo laboral.
Si no lo es: se regenera **una vez** con el motivo como feedback; si vuelve a fallar, se devuelve un mensaje fijo.

````python
FALLBACK = "Prefiero centrarme en mi trayectoria profesional. ¿Quieres saber algo sobre mi experiencia, proyectos o habilidades?"

def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)

    while response.choices[0].finish_reason == "tool_calls":
        message = response.choices[0].message
        messages.append(message)
        for tool_call in message.tool_calls:
            email = json.loads(tool_call.function.arguments).get("email")
            record_email_tool(email)
            messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
        response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)

    reply = response.choices[0].message.content
    verdict = evaluate(reply)
    print("Evaluación:", verdict)

    if not verdict["is_work_related"]:
        messages.append({"role": "assistant", "content": reply})
        messages.append({"role": "user", "content": f"Tu respuesta no es estrictamente laboral ({verdict['reason']}). Reescríbela centrada solo en tu trayectoria profesional."})
        reply = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages).choices[0].message.content
        verdict = evaluate(reply)
        print("Reintento:", verdict)
        if not verdict["is_work_related"]:
            reply = FALLBACK

    return reply
````

````python
gr.ChatInterface(chat).launch(inbrowser=True)
````
